# 📓 Notebook 4｜類別可分離度量：散佈矩陣 J3（站 4）

> 對應講義 Part 4（5.6）。給一個「數字」衡量兩類分得多開。
> 散佈矩陣是幾何版、好算：類內散佈 $S_w$、類間散佈 $S_b$、混合散佈 $S_m=S_w+S_b$。
> 準則 $J_3=\operatorname{trace}\{S_w^{-1}S_m\}$。

```
import numpy as np
rng = np.random.default_rng(0)

def scatter_J3(X1, X2):
    '''兩類樣本 X1, X2（各 N×d），算 J3 = trace(Sw^{-1} Sm)。'''
    mu1, mu2 = X1.mean(0), X2.mean(0)
    mu0 = (mu1 + mu2) / 2
    Sw = np.cov(X1.T) + np.cov(X2.T)          # 類內散佈（等權）
    Sm = Sw + (mu1 - mu0)[:, None] @ (mu1 - mu0)[None, :]             + (mu2 - mu0)[:, None] @ (mu2 - mu0)[None, :]
    return np.trace(np.linalg.inv(Sw) @ Sm)

# 課本 Figure 5.5 三種分布（描述對照，非精確資料）
# (a) 類內緊、類間近  (b) 類內散、類間近  (c) 類內緊、類間遠
cases = {
    'a 類內緊·類間近': (np.array([0.0, 0.0]), np.array([0.5, 0.0]), 0.3),
    'b 類內散·類間近': (np.array([0.0, 0.0]), np.array([0.5, 0.0]), 2.0),
    'c 類內緊·類間遠': (np.array([0.0, 0.0]), np.array([3.0, 0.0]), 0.3),
}
for name, (m1, m2, sd) in cases.items():
    X1 = rng.normal(m1, sd, (200, 2)); X2 = rng.normal(m2, sd, (200, 2))
    print(f'{name}: J3 = {scatter_J3(X1, X2):.1f}')
print('課本 Example 5.5: J3 = 164.7 / 12.5 / 620.9（a/b/c）')
print('→ 排序一致：類間遠、類內緊的 c 最大；類間近、類內散的 b 最小')
```

### Fisher 判別比 FDR（一維特徵的簡化版）

一維兩類時，$FDR=\frac{(\mu_1-\mu_2)^2}{\sigma_1^2+\sigma_2^2}$。

```
def fdr(x1, x2):
    return (x1.mean() - x2.mean()) ** 2 / (x1.var() + x2.var())

# 用 NB2 的 Example 5.3 資料
w1 = np.array([3.5, 3.7, 3.9, 4.1, 3.4, 3.5, 4.1, 3.8, 3.6, 3.7])
w2 = np.array([3.2, 3.6, 3.1, 3.4, 3.0, 3.4, 2.8, 3.1, 3.3, 3.6])
print(f'FDR = {fdr(w1, w2):.3f}')

# ✏️ 練習：把 w2 全部 +0.5，看 FDR 如何變化
```
