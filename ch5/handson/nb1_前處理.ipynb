# 📓 Notebook 1｜前處理：離群值・正規化・缺值（站 7）

> 對應講義 Part 7（5.2）。選特徵前先「洗菜」：踢離群值、統一尺度、補缺值。

```
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(0)

# 合成一個特徵：正常值 + 幾個離群值
x = np.concatenate([rng.normal(0, 1, 100), [8.0, -7.5, 9.2]])
print('原始樣本數:', len(x), ' 平均:', round(x.mean(), 2), ' 標準差:', round(x.std(), 2))
```

### 離群值移除：用「中位數 ± k×MAD」或「z-score 閾值」

離群值會把平均數/變異數拉歪，進而誤導後續的 t-test 與散佈矩陣。

```
# 方法：z-score 絕對值 > 3 視為離群值
z = (x - x.mean()) / x.std()
mask = np.abs(z) < 3
x_clean = x[mask]
print('移除離群值後樣本數:', len(x_clean), ' 平均:', round(x_clean.mean(), 2), ' 標準差:', round(x_clean.std(), 2))

plt.figure(figsize=(7, 3))
plt.subplot(1, 2, 1); plt.hist(x, bins=30); plt.title('原始（有離群值）')
plt.subplot(1, 2, 2); plt.hist(x_clean, bins=30); plt.title('移除後')
plt.tight_layout(); plt.show()
```

### 正規化：把不同尺度的特徵拉到同一範圍

兩種常用：**min-max**（壓到 [0,1]）與 **z-score**（減平均除標準差）。

```
# min-max 與 z-score
x_mm = (x_clean - x_clean.min()) / (x_clean.max() - x_clean.min())
x_zs = (x_clean - x_clean.mean()) / x_clean.std()
print('min-max 範圍:', round(x_mm.min(), 3), '~', round(x_mm.max(), 3))
print('z-score 平均:', round(x_zs.mean(), 3), ' 標準差:', round(x_zs.std(), 3))
```

### 缺值填補：平均 / 中位數 / 最常見值

缺值（NaN）會讓很多演算法直接炸掉，要先補。

```
# 製造缺值，再用中位數填補
x_miss = x_clean.copy()
x_miss[rng.choice(len(x_miss), 10, replace=False)] = np.nan
print('缺值數:', np.isnan(x_miss).sum())

med = np.nanmedian(x_miss)          # 中位數（忽略 NaN）
x_fill = np.where(np.isnan(x_miss), med, x_miss)
print('填補後缺值數:', np.isnan(x_fill).sum(), ' 填補值(中位數):', round(med, 3))

# ✏️ 練習：改用平均數填補，比較結果
```
